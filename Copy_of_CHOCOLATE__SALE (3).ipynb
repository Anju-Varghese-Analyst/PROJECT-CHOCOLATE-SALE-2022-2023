**Step** 4 ** -**Handle the missing values using suitable methods.**

Missing values were identified in four columns: Order_Date, Discount_Percentage, Price_per_Box, and Marketing_Spend.

To handle the missing values appropriately, we analyze the nature of each column. Since different columns contain different types of data (dates, financial metrics, marketing metrics), we use specific methods for each rather than applying a single method to all.

1.** Order_Date**

* **Missing Values Count**: 437 missing rows

* **Handling Method Applied**: Remove missing rows

* **Justification** :
To handle missing values in the order_date column of a large sales dataset without sorting the chronological order.
The dataset contains 200,000 rows, out of which the order_date column has 484 missing values. Since the dataset is not in chronological order, sequential imputation techniques like Forward Fill (ffill) cannot be safely applied.
Since the missing values account for only 0.24% of the entire dataset, removing these rows will not cause any significant information loss or skew the data analysis. Therefore, Listwise Deletion (Row Dropping) via the dropna() method is selected as the safest and most efficient solution.

**CONCLUSION**: The 437 rows with missing dates were successfully dropped from the dataset. The total row count changed from 200,000 to 199,563, leaving a clean dataset ready for subsequent analysis without introducing any incorrect or assumed dates.

**2. Discount_Pct**

* **Missing Values Count** : 489 missing rows

* **Handling Method Applied**: Product-wise Median Imputation

* **Justification**: A global median across all rows would be highly inaccurate because different items have entirely different profit margins and promotional strategies. By calculating the median discount grouped by each specific Product, we ensure that the filled discount rate aligns accurately with the historical discount behavior of that particular item.

3.**Price_per_Box**

* **Missing Values Count**: 457 missing rows

* **Handling Method Applied**: Product-wise Median Imputation

* **Justification**: Box prices vary drastically depending on the item type (e.g., premium products vs. economy items). Applying an overall database average would heavily distort the revenue calculations. Grouping by Product and replacing the missing cells with that item's specific median price ensures the pricing records remain realistic.

4.** Marketing_Spend**

* **Missing Values Count**: 461 missing rows

* **Handling Method Applied**: Overall Median Imputation

* **Justification**: Marketing expenses represent overall operational campaigns rather than individual transactional attributes tied to a single product line. Therefore, filling the missing gaps with the overall median of the entire column is appropriate. Using the median instead of the mean prevents extreme high or low spending spikes (outliers) from skewing the data distribution.

```python
df['Order_Date'].isna().sum()

```

```python
# Step 2: Drop the 484 rows where 'order_date' contains missing values (NaN/NaT)
df= df.dropna(subset=['Order_Date'])
```

```python
# Find the median value of Marketing_Spend
median_marketing = df['Marketing_Spend'].median()

# Replace missing values with the median
df['Marketing_Spend'] = df['Marketing_Spend'].fillna(median_marketing)
```

The maximum sales amount is 12,138.09, which is much higher than the typical sales values. Typical Range: Most of  sales transactions (the middle 50%) fall between $287.38 and $622.66.

The Sales_Amount column was analyzed to identify potential outliers. The Interquartile Range (IQR) method was used for this analysis. Q1 and Q3 were calculated, and the IQR was obtained by subtracting Q1 from Q3. The lower and upper bounds were then calculated using the 1.5 × IQR rule. Sales amount values falling outside these bounds were considered potential outliers.

Purpose:

This analysis helps to identify unusually high or low sales amount values in the dataset.

```python
# Calculate the first quartile (Q1)
Q1 = df['Sales_Amount'].quantile(0.25)

# Calculate the third quartile (Q3)
Q3 = df['Sales_Amount'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Calculate the lower bound for detecting outliers
lower_bound = Q1 - 1.5 * IQR

# Calculate the upper bound for detecting outliers
upper_bound = Q3 + 1.5 * IQR

# Display the calculated values
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Identify potential outliers in Sales_Amount
outliers = df[(df['Sales_Amount'] < lower_bound) |
              (df['Sales_Amount'] > upper_bound)]

# Count the number of potential outliers
print("Number of potential outliers:", len(outliers))
```

Conclusion:

The Sales_Amount column was analyzed using the IQR method to identify potential outliers. The first quartile (Q1) was 287.34 and the third quartile (Q3) was 622.50. The IQR was 335.16, and the upper bound was approximately 1,125.24. A total of 10,750 values were identified as potential outliers. These values were not removed because they may represent genuine high-value sales transactions.

Any value that is greater than \(Q3 + (1.5 \times IQR)\) is considered a high outlier. Let's calculate your upper limit:

1.Upper Limit = 181 + (1.5 * 112) = 181 + 168 = 349


Since your maximum value (3,811) is much higher than 349, it is officially an outlier.

1. **Monthly Sales Trend**

A. Why is this visualization used?

This visualization is used to understand how sales change from month to month. It helps to identify periods of  higher and lower sales and to observe the overall sales trend over time.

B. What analysis is being performed?

This is a time-based analysis.
The Order_Date column is used to identify the year and month of each sale. Then, the Amount column is grouped by month and the total sales amount for each month is calculated.
A line chart is used because it makes it easy to observe changes and trends in sales over time.

C. Steps performed before writing the code

Before creating the visualization, I first checked the Order_Date column and made sure that it was converted into the correct datetime data type.
Then, I extracted the year and month from the Order_Date column. After that, I grouped the data month-wise and calculated the total sales amount for each month.

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
df=df.copy()
df['Month'] = df['Order_Date'].dt.to_period('M')
# Extract the year and month from the Order_Date column

monthly_sales = df.groupby('Month')['Sales_Amount'].sum()
# Group the data by month
# Calculate the total sales amount for each month

plt.figure(figsize=(12, 6))
# Set the size of the chart

plt.plot(monthly_sales.index.astype(str), monthly_sales.values,color='#1A5276',linewidth=2.5,marker='o',markersize=5,markerfacecolor="#E74C3C")
# Create a line chart using month and total sales amount

plt.title('Monthly Sales Trend',fontsize=16,fontweight='bold',pad=15,color='#2C3E50')
# Add a title to the chart

plt.xlabel('Month',fontsize=12,labelpad=10)
# Add a label to the X-axis

plt.ylabel('Total Sales Amount',fontsize=12,labelpad=10)
# Add a label to the Y-axis

plt.xticks(rotation=45,ha='right')
# Rotate the month labels for better readability
# Function to convert large raw numbers into a clean millions format (e.g., 10000000 -> 10.0M)
def millions_formatter(x, pos):
    return f'{x*1e-6:.1f}M'

# Apply the millions_formatter function to the Y-axis labels of the current plot
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(millions_formatter))
# Get Current Axes (plt.gca) to customize and format the Y-axis number labels

plt.tight_layout()
# Adjust the layout of the chart

plt.show()
# Display the chart
```

D.What can be understood from the graph?

The line chart shows that the total sales amount changes across different months. The upward movements indicate an increase in sales, while the downward movements indicate a decrease in sales. This helps to identify the monthly sales pattern and periods of relatively higher and lower sales.

E. Insights from the visualization

The monthly sales trend shows fluctuations in sales across the observed period. Sales were relatively high in January 2022 and July 2023, while lower sales were observed around December 2022. From January 2023 onwards, the sales level remained generally higher than most of the months in 2022. Overall, the graph shows that sales vary from month to month rather than following a constant trend.
